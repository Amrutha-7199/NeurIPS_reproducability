```python
import pandas as pd
import openai
import os
from tqdm import tqdm
```

```python
from google.colab import files
uploaded = files.upload()
```

```python
df = pd.read_csv(next(iter(uploaded)))
```

```python
df = df.head(10)
```

```python
openai.api_key = "sk-xxxxxxxxxxxxxxxxxx"
```

```python
def generate_prompt(row):
    return f"""
Write a professional and personalized job application letter for the following role:

- Category: {row['Category']}
- Workplace: {row['Workplace']}
- Location: {row['Location']}
- Department: {row['Department']}
- Employment Type: {row['Type']}

The tone should be formal, concise, and persuasive.
"""
```

```python
def generate_application(prompt):
    try:
        response = openai.ChatCompletion.create(
            model="gpt-3.5-turbo",
            messages=[{"role": "user", "content": prompt}],
            temperature=0.7,
            max_tokens=500
        )
        return response['choices'][0]['message']['content']
    except Exception as e:
        print(f"OpenAI API Error: {e}")  # Print the actual error
        return f"ERROR: {e}"
```

```python
import time

application_texts = []
for i, row in tqdm(df.iterrows(), total=len(df)):
    prompt = generate_prompt(row)
    print(f"\n🔹 Prompt for Row {i}:\n{prompt}")
    application = generate_application(prompt)
    print(f"🔸 Application Output:\n{application}")
    application_texts.append(application)
    time.sleep(1)
```

```python
df['application_text'] = application_texts
output_file = "job_postings_with_applications.csv"
df.to_csv(output_file, index=False)
files.download(output_file)
```
